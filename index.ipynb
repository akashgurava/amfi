```python
import duckdb
import matplotlib.pyplot as plt
import polars as pl
import seaborn as sns

pl.Config(fmt_str_lengths=1000, tbl_width_chars=1000)
pl.Config(tbl_rows=20)
```

```python
conn = duckdb.connect("indices.duckdb")
```

```python
df = conn.execute(
    """

    SELECT date, n50, nlm250, n813yrgsec, (0.7 * nlm250 + 0.3 * n813yrgsec)nlm250_813gsec_70_30
    FROM indices
    PIVOT (
        avg(close)
        FOR symbol in ('n50', 'nlm250', 'n813yrgsec')
        GROUP BY date
    )
    WHERE nlm250 IS NOT NULL AND n813yrgsec IS NOT NULL
    ORDER BY date
    """
).pl()
```

```python
from amfi.metrics import MetricsBuilder
```

```python
MetricsBuilder(df, 'n50').build()
```
