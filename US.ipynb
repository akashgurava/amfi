```python
import polars as pl
import duckdb
pl.Config(fmt_str_lengths=1000, tbl_width_chars=1000)
pl.Config(tbl_rows=10)
```

```python
conn = duckdb.connect()
```

```python
conn.execute("""
    SELECT *
    FROM 'out/joined.csv'
    WHERE TRUE
        AND returns_trailing_6m > -0.06
        AND returns_trailing_1y > -0.06
        AND returns_trailing_3y > -0.06
        AND returns_trailing_5y > -0.06

        AND returns_yearly_2024 > -0.06
        AND returns_yearly_2020 > -0.06

        AND cagr_trailing_6m > -0.06
        AND cagr_trailing_1y > -0.06
        AND cagr_trailing_3y > -0.06
        AND cagr_trailing_5y > -0.06

        AND max_dd_trailing_6m > -0.4
        AND max_dd_trailing_1y > -0.4
        AND max_dd_trailing_3y > -0.4
        AND max_dd_trailing_5y > -0.4

        AND max_dd_yearly_2024 > -0.4
        AND max_dd_yearly_2020 > -0.4

    """
).pl()
```

```python
conn.execute("""
    SELECT *
    FROM 'out/joined.csv'
    WHERE TRUE
        AND returns_trailing_5y >= 0.5
        AND returns_trailing_3y >= 0.3

        AND cagr_trailing_5y >= 0.1
        AND cagr_trailing_3y >= 0.1
    """
).pl()
```

```python
conn.execute("""
    SELECT ticker, c1y.*
    FROM 'out/joined.csv' j
    LEFT JOIN 'out/trailing_1y_corr.csv' c1y USING (ticker)
    WHERE TRUE
        AND returns_trailing_5y >= 0.5
        AND returns_trailing_3y >= 0.3

        AND returns_yearly_2025 >= 0.1
        AND returns_yearly_2024 >= 0.1
        AND returns_yearly_2017 >= 0.1

        AND cagr_trailing_5y >= 0.1
        AND cagr_trailing_3y >= 0.1
    """
).pl()
```

```python
conn.execute("""
    WITH filtered_tickers AS (
    -- 1. Get the base tickers meeting your performance criteria
    SELECT ticker, cagr_trailing_5y
    FROM 'out/joined.csv'
    WHERE TRUE
        AND ticker NOT IN (
        'UGL', 'NUGT', 'RING', 'GDX', 'GOEX', 'SGDM'
    )
),
all_target_cagrs AS (
    -- 2. Extract CAGRs for the target tickers so we can rank them
    SELECT ticker, cagr_trailing_5y AS target_cagr
    FROM 'out/joined.csv'
    WHERE TRUE
        AND ticker NOT IN (
        'UGL', 'NUGT', 'RING', 'GDX', 'GOEX', 'SGDM', 'AGQ', 'GDXJ', 'OUNZ', 'USCI', 'SDCI', 'JNUG',
        'FENY', 'UGA', 'RSPG', 'MLPX', 'VDE', 'UMI', 'BNO', 'USO', 'ENFR', 'IXC'
    )
    
),
unpivoted_corr AS (
    -- 3. Convert the wide correlation matrix into a long format
    SELECT 
        ticker AS base_ticker, 
        target_ticker, 
        CAST(corr_value AS DOUBLE) AS corr_value
    FROM 'out/trailing_1y_corr.csv'
    UNPIVOT (corr_value FOR target_ticker IN (COLUMNS(* EXCLUDE (ticker))))
),
ranked_corr AS (
    -- 4. Join base stats, target stats, and filter for corr < 0.2
    SELECT 
        u.base_ticker,
        f.cagr_trailing_5y AS base_cagr,
        u.target_ticker,
        u.corr_value,
        t.target_cagr,
        -- Rank the target tickers by their 5Y CAGR (Highest first)
        ROW_NUMBER() OVER (
            PARTITION BY u.base_ticker 
            ORDER BY t.target_cagr DESC
        ) as rn_low_corr
    FROM unpivoted_corr u
    JOIN filtered_tickers f ON u.base_ticker = f.ticker
    JOIN all_target_cagrs t ON u.target_ticker = t.ticker
    WHERE u.base_ticker != u.target_ticker  -- Ignore self-correlation
      AND u.corr_value IS NOT NULL
      AND u.corr_value < 0.2                -- Apply the correlation threshold
)
-- 5. Pivot the top 5 highest-CAGR targets back into the requested format
SELECT 
    base_ticker AS ticker, 
    MAX(base_cagr) AS cagr_trailing_5y,
    
    MAX(CASE WHEN rn_low_corr = 1 THEN target_ticker END) AS low_corr_ticker_t1y_1,
    MAX(CASE WHEN rn_low_corr = 1 THEN corr_value END) AS low_corr_t1y_1,
    
    MAX(CASE WHEN rn_low_corr = 2 THEN target_ticker END) AS low_corr_ticker_t1y_2,
    MAX(CASE WHEN rn_low_corr = 2 THEN corr_value END) AS low_corr_t1y_2,
    
    MAX(CASE WHEN rn_low_corr = 3 THEN target_ticker END) AS low_corr_ticker_t1y_3,
    MAX(CASE WHEN rn_low_corr = 3 THEN corr_value END) AS low_corr_t1y_3,
    
    MAX(CASE WHEN rn_low_corr = 4 THEN target_ticker END) AS low_corr_ticker_t1y_4,
    MAX(CASE WHEN rn_low_corr = 4 THEN corr_value END) AS low_corr_t1y_4,
    
    MAX(CASE WHEN rn_low_corr = 5 THEN target_ticker END) AS low_corr_ticker_t1y_5,
    MAX(CASE WHEN rn_low_corr = 5 THEN corr_value END) AS low_corr_t1y_5

FROM ranked_corr
GROUP BY base_ticker
ORDER BY cagr_trailing_5y DESC;
    """
).pl()
```
